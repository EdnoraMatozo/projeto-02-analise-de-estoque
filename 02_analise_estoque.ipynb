# Projeto 02 – Análise de Estoque

Este projeto tem como objetivo analisar os dados de estoque de uma empresa, identificando produtos com baixo estoque, itens que precisam de reposição e mercadorias com quantidade excessiva.

A análise será realizada utilizando Python, Pandas e ferramentas de visualização de dados.

## Objetivos

* Calcular o valor total de cada produto em estoque;
* Identificar produtos com estoque abaixo do mínimo;
* Identificar produtos com excesso de estoque;
* Analisar o estoque por categoria;
* Criar gráficos para facilitar a interpretação dos resultados;
* Apresentar conclusões e recomendações com base nos dados.

```python
# Quantidade de produtos por categoria
categoria = df["Category"].value_counts()

print(categoria)
```

```python
import matplotlib.pyplot as plt

categoria.plot(kind="bar")

plt.title("Quantidade de produtos por categoria")
plt.xlabel("Categoria")
plt.ylabel("Quantidade")
plt.xticks(rotation=45)

plt.show()
```

```python
# Contagem de produtos por categoria
categoria = df["Category"].value_counts()

# Cria o gráfico de pizza
plt.figure(figsize=(6, 6))

plt.pie(
    categoria,
    labels=categoria.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribuição dos produtos por categoria")

plt.show()
```

```python
# Histograma dos níveis de estoque

plt.figure(figsize=(8,6))

plt.hist(df["Inventory Level"], bins=20)

plt.title("Distribuição dos níveis de estoque")
plt.xlabel("Nível de estoque")
plt.ylabel("Quantidade de produtos")

plt.show()
```

```python
resumo_estoque = df.groupby('Product ID').agg({
    'Inventory Level': 'sum'
}).sort_values('Inventory Level')

resumo_estoque
```

```python
# Products with critically low inventory
critical_stock = df[df["Inventory Level"] <= 50]

critical_stock.head()
```

```python
df["Inventory Level"].min()
```

```python
critical_stock.shape[0]
```

```python
critical_stock["Category"].value_counts()
```

```python
# Critical stock by category
critical_by_category = critical_stock["Category"].value_counts()

plt.figure(figsize=(8, 5))
critical_by_category.plot(
    kind="bar",
    color=["royalblue", "orange", "seagreen", "mediumpurple", "crimson"]
)

plt.title("Critical Stock by Category")
plt.xlabel("Category")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Conclusão da análise

A análise permitiu identificar os produtos com níveis críticos de estoque e comparar sua distribuição entre as diferentes categorias. Essas informações podem auxiliar no planejamento de reposição, priorizando produtos e categorias com maior necessidade de reabastecimento.
